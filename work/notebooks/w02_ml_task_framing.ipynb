## 4. The unit of analysis, as a real dataframe

*Load your lane's slice and show it: one row = one what?*

```python

import pandas as pd
import numpy as np

# Unit of Analysis: One row = One Unique Content Page / URL
data = {
    'page_id': [101, 102, 103, 104, 105],
    'word_count': [1200, 450, 2100, 800, 1500],
    'backlinks': [15, 2, 45, 0, 12],
    'avg_time_sec': [110, 35, 180, 40, 95],
    'page_views': [2500, 300, 5400, 150, 1800]
}

df = pd.DataFrame(data)

# Target Column Creation
df['is_high_performing'] = np.where((df['page_views'] > 1000) & (df['avg_time_sec'] > 90), 1, 0)

print("Unit of Analysis: One row represents one Content URL")
df
```

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

**Why ML Beats Fixed Rules:**
1. **Complex Interactions:** Simple IF-THEN rules miss non-linear feature relationships (e.g., high backlinks can compensate for lower word count).
2. **Dynamic Adaptation:** Search engine dynamics change over time; an ML model adapts with new training data, whereas fixed rules break.
