## 1. Ranked actions + reason codes

*The queue: what to do first, and why, in words a human trusts.*

```python
import pandas as pd
import numpy as np
import os

# Create outputs directory if it doesn't exist
os.makedirs('../outputs', exist_ok=True)
os.makedirs('../figures', exist_ok=True)

# Generate dummy score data for playbook mapping
np.random.seed(42)
urls = [f"https://example.com/page-{i}" for i in range(1, 51)]
scores = np.random.uniform(0.1, 0.95, 50)
impressions_decay = np.random.uniform(-0.6, 0.1, 50)

df = pd.DataFrame({
    'url': urls,
    'model_score': scores,
    'impression_decay_rate': impressions_decay
})

# Archetype and Action Mapping Logic
def map_action(row):
    if row['model_score'] > 0.8 and row['impression_decay_rate'] < -0.3:
        return 'Urgent Refresh', 'RC_HIGH_DECAY', 'High priority content decay detected with strong model confidence.'
    elif row['model_score'] > 0.6:
        return 'Content Optimize', 'RC_MODERATE_DROP', 'Moderate performance drop; title or meta refresh recommended.'
    elif row['model_score'] < 0.3:
        return 'Monitor', 'RC_STABLE', 'Traffic stable or low impact; continue passive monitoring.'
    else:
        return 'Internal Link Boost', 'RC_LOW_AUTHORITY', 'Page performance lagging due to potential link structure issues.'

df[['recommended_action', 'reason_code', 'reason_description']] = df.apply(map_action, axis=1, result_type='expand')

# Rank the queue
ranked_queue = df.sort_values(by='model_score', ascending=False).reset_index(drop=True)
print("Top Ranked Action Items:")
print(ranked_queue[['url', 'model_score', 'recommended_action', 'reason_code']].head())
```

## 3. Human review + the no-go list

*What a person must check before acting. What should never be automated.*

# 3. Human Review Rules & No-Go List

### Mandatory Human Review Rules
1. **High Impact Review:** Any URL with high commercial intent or primary conversion landing pages must undergo manual editorial review before executing a rewrite.
2. **Sanity Check:** Content editors must verify if traffic drop is due to intentional canonicalization or site migration before flagging as decay.

### The No-Go List (DO NOT Automate)
* **Automated Page Deletion:** Never automatically delete or redirect 404/410 pages based solely on model scores.
* **Auto-Rewriting Core Content:** Never auto-generate and publish AI content directly to live production URLs without human verification.
* **Legal/Compliance Pages:** Exclude terms of service, privacy policies, and medical/financial advice pages completely from automated recommendation queues.
