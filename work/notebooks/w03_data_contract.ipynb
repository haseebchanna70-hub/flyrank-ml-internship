## 3. Verify it with queries (grain, counts, missing values, windows)

*Every claim above gets a query cell here. A contract claim without a query next to it is a guess.*

```python
import os
from google.colab import userdata
import pandas as pd
from datasets import load_dataset
import duckdb

# 1. Fetch HF Token from Colab Secrets
hf_token = userdata.get('HF_TOKEN')

# 2. Load the correct daily performance config from Hugging Face
print("Loading dataset config from Hugging Face...")
dataset = load_dataset("FlyRank/internship-warehouse", "fact_content_daily_performance", token=hf_token)

# Convert train split to pandas dataframe
df = dataset['train'].to_pandas()

# Connect to DuckDB locally using the dataframe
con = duckdb.connect()
con.register('daily_url_performance', df)

print("\n--- Query 1: Verify Grain (One row = url_id + date) ---")
q1 = """
SELECT url_id, date, COUNT(*) as row_count
FROM daily_url_performance
WHERE date BETWEEN '2026-03-01' AND '2026-03-31'
GROUP BY url_id, date
HAVING COUNT(*) > 1;
"""
df_q1 = con.execute(q1).df()
print("Duplicate Rows Count (0 means verified):", len(df_q1))

print("\n--- Query 2: Date Span & Row Count ---")
q2 = """
SELECT
    MIN(date) as min_date,
    MAX(date) as max_date,
    COUNT(*) as total_rows,
    COUNT(DISTINCT url_id) as unique_urls
FROM daily_url_performance
WHERE date BETWEEN '2026-03-01' AND '2026-03-31';
"""
print(con.execute(q2).df())

print("\n--- Query 3: Availability Filter (IS TRUE) ---")
q3 = """
SELECT
    COUNT(*) as total_rows,
    COUNT(CASE WHEN is_indexable IS TRUE THEN 1 END) as active_rows
FROM daily_url_performance
WHERE date BETWEEN '2026-03-01' AND '2026-03-31';
"""
print(con.execute(q3).df())
```
